```python
# ============================================
# SafeSpace AI — Content Moderator
# Version 0.1
# ============================================

# 1. IMPORT LIBRARIES
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix


# ============================================
# 2. CREATE A SMALL STARTER DATASET
# ============================================

data = {
    "text": [
        "Have a wonderful day",
        "I completely disagree with you",
        "You are worthless",
        "Nobody wants you here",
        "Someone should find her address",
        "I know where she lives",
        "Someone should hurt her",
        "You deserve to be attacked",
        "Women like you should not speak",
        "Women are inferior to men",
        "You are disgusting",
        "Stop contacting me"
    ],

    "label": [
        "normal",
        "normal",
        "harassment",
        "harassment",
        "doxing",
        "doxing",
        "threat",
        "threat",
        "gender_hate",
        "gender_hate",
        "harassment",
        "normal"
    ]
}

df = pd.DataFrame(data)

print(df)
print(df["label"].value_counts())
df.head()
x = df["text"]
y = df["label"]
x_train, x_test,y_train ,y_test = train_test_split(
    x,
    y,
    test_size=0.5, # Increased test_size to ensure enough samples per class for stratification
    random_state=42,
    stratify=y
    )
print("Training examples:",
len(x_train))
print("Testing examples:",
len(x_test))
vectorizer = TfidfVectorizer()

X_train_tfidf = vectorizer.fit_transform(x_train) # Corrected X_train to x_train
X_test_tfidf = vectorizer.transform(x_test) # Corrected X_test to x_test

print(X_train_tfidf.shape)
model = LogisticRegression(max_iter=1000)

model.fit(X_train_tfidf, y_train)
# The following two lines are redundant, as the model was already initialized and fitted.
# model = LogisticRegression(max_iter=1000)
# model.fit(X_train_tfidf, y_train)
new_message = [
    "Someone should find out where she lives"
]

new_message_tfidf = vectorizer.transform(new_message)

prediction = model.predict(new_message_tfidf)

print("Prediction:", prediction[0])
```
